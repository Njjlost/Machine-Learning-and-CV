```python
import io
import numpy as np
import matplotlib.pyplot as plt
import cv2
from skimage.feature import local_binary_pattern
from scipy.spatial.distance import pdist, squareform
import seaborn as sns
from google.colab import files
from PIL import Image

# ==========================================
# 1. UPLOAD IMAGE IN GOOGLE COLAB
# ==========================================
print("Please upload your image file:")
uploaded = files.upload()

# Grab the first uploaded file name
file_name = list(uploaded.keys())[0]

# Load uploaded image using PIL and convert to RGB
img_raw = Image.open(io.BytesIO(uploaded[file_name])).convert('RGB')
img_rgb = np.array(img_raw)

# Optional: Resize to a standard maximum dimension for clean visualization & fast execution
max_dim = 900
h_raw, w_raw, _ = img_rgb.shape
if max(h_raw, w_raw) > max_dim:
    scale = max_dim / max(h_raw, w_raw)
    img_rgb = cv2.resize(img_rgb, (int(w_raw * scale), int(h_raw * scale)))

# ==========================================
# 2. PATCH EXTRACTION & FEATURE COMPUTATION
# ==========================================
GRID_ROWS, GRID_COLS = 3, 3  # Adjust grid dimensions here if needed
h, w, c = img_rgb.shape
patch_h, patch_w = h // GRID_ROWS, w // GRID_COLS

patches = []
features_list = []

for r in range(GRID_ROWS):
    for c_idx in range(GRID_COLS):
        # Extract patch boundaries
        y_start, y_end = r * patch_h, (r + 1) * patch_h
        x_start, x_end = c_idx * patch_w, (c_idx + 1) * patch_w
        patch = img_rgb[y_start:y_end, x_start:x_end]
        patches.append(patch)

        # 1. RGB Color Histogram Descriptors
        hist_r, _ = np.histogram(patch[:, :, 0], bins=16, range=(0, 256), density=True)
        hist_g, _ = np.histogram(patch[:, :, 1], bins=16, range=(0, 256), density=True)
        hist_b, _ = np.histogram(patch[:, :, 2], bins=16, range=(0, 256), density=True)
        color_hist = np.concatenate([hist_r, hist_g, hist_b])

        # 2. Local Binary Pattern (LBP) Texture Descriptors
        gray_patch = cv2.cvtColor(patch, cv2.COLOR_RGB2GRAY)
        lbp = local_binary_pattern(gray_patch, P=8, R=1, method="uniform")
        lbp_hist, _ = np.histogram(lbp.ravel(), bins=10, range=(0, 10), density=True)

        # Combine vectors for feature representation
        feature_vector = np.concatenate([color_hist, lbp_hist])
        features_list.append(feature_vector)

features_array = np.array(features_list)

# ==========================================
# 3. VISUALIZE PATCHES & HISTOGRAMS
# ==========================================
plt.figure(figsize=(14, 10))
for i in range(len(patches)):
    # Display Patch Segment
    plt.subplot(GRID_ROWS * 2, GRID_COLS, i + 1 + (i // GRID_COLS) * GRID_COLS)
    plt.imshow(patches[i])
    plt.title(f"Patch {i}", fontsize=10)
    plt.axis("off")

    # Display Color Histogram per Segment
    plt.subplot(GRID_ROWS * 2, GRID_COLS, i + 1 + (i // GRID_COLS + 1) * GRID_COLS)
    for color, channel in zip(('red', 'green', 'blue'), (0, 1, 2)):
        hist, _ = np.histogram(patches[i][:, :, channel], bins=32, range=(0, 256))
        plt.plot(hist, color=color, alpha=0.7)
    plt.title(f"Hist P{i}", fontsize=9)
    plt.xlim([0, 32])

plt.tight_layout()
plt.suptitle("Uploaded Image: Patches & Color Histograms", y=1.02, fontsize=16)
plt.show()

# ==========================================
# 4. METRICS & HEATMAP VISUALIZATION
# ==========================================
# Euclidean Distance Matrix (0 = identical, higher = more distant)
euclidean_dist_matrix = squareform(pdist(features_array, metric='euclidean'))

# Pearson Correlation Matrix & Dissimilarity (1 - Correlation)
correlation_matrix = np.corrcoef(features_array)
correlation_dissimilarity = 1 - correlation_matrix

fig, axes = plt.subplots(1, 2, figsize=(15, 6))
labels = [f"P{i}" for i in range(len(patches))]

# Heatmap 1: Euclidean Distance
sns.heatmap(
    euclidean_dist_matrix,
    annot=True,
    fmt=".2f",
    cmap="YlOrRd",
    ax=axes[0],
    xticklabels=labels,
    yticklabels=labels
)
axes[0].set_title("Euclidean Distance Matrix (Lower = More Similar)")

# Heatmap 2: Correlation Dissimilarity
sns.heatmap(
    correlation_dissimilarity,
    annot=True,
    fmt=".2f",
    cmap="Blues",
    ax=axes[1],
    xticklabels=labels,
    yticklabels=labels
)
axes[1].set_title("Correlation Dissimilarity Matrix (1 - Correlation)")

plt.tight_layout()
plt.show()
```

```python
import io
import cv2
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
from scipy.spatial.distance import pdist, squareform
from skimage.feature import local_binary_pattern
from google.colab import files

# Upload Image File
uploaded = files.upload()
file_name = list(uploaded.keys())[0]



g_raw = Image.open(io.BytesIO(uploaded[file_name])).convert('RGB')
img_rgb = np.array(img_raw)

max_dim = 900
h_raw, w_raw, _ = img_rgb.shape
if max(h_raw, w_raw) > max_dim:
    scale = max_dim / max(h_raw, w_raw)
    img_rgb = cv2.resize(img_rgb, (int(w_raw * scale), int(h_raw * scale)))

h, w, c = img_rgb.shape
GOLDEN_RATIO = 1.61803398875
phi_inv = 1 / GOLDEN_RATIO  # ~0.618


golden_boxes = [
    ("P0: Global (Full)", 0.0, 1.0, 0.0, 1.0),
    ("P1: Golden Cut Y", 0.0, phi_inv, 0.0, 1.0),
    ("P2: Golden Cut X", 0.0, 1.0, 0.0, phi_inv),
    ("P3: Center Focus", (1 - phi_inv) / 2, (1 + phi_inv) / 2, (1 - phi_inv) / 2, (1 + phi_inv) / 2),
    ("P4: Top-Left Golden", 0.0, phi_inv, 0.0, phi_inv),
    ("P5: Bottom-Right Local", 1 - phi_inv, 1.0, 1 - phi_inv, 1.0)
]

patches = []
patch_labels = []
features_list = []

for label, ys, ye, xs, xe in golden_boxes:
    y_start, y_end = int(ys * h), int(ye * h)
    x_start, x_end = int(xs * w), int(xe * w)

    patch = img_rgb[y_start:y_end, x_start:x_end]
    patches.append(patch)
    patch_labels.append(label)

    hist_r, _ = np.histogram(patch[:, :, 0], bins=16, range=(0, 256), density=True)
    hist_g, _ = np.histogram(patch[:, :, 1], bins=16, range=(0, 256), density=True)
    hist_b, _ = np.histogram(patch[:, :, 2], bins=16, range=(0, 256), density=True)
    color_hist = np.concatenate([hist_r, hist_g, hist_b])

    gray_patch = cv2.cvtColor(patch, cv2.COLOR_RGB2GRAY)
    lbp = local_binary_pattern(gray_patch, P=8, R=1, method="uniform")
    lbp_hist, _ = np.histogram(lbp.ravel(), bins=10, range=(0, 10), density=True)


    feature_vector = np.concatenate([color_hist, lbp_hist])
    features_list.append(feature_vector)

features_array = np.array(features_list)

# Visualization: Patches & RGB Histograms Layout (2 Rows x 3 Cols per sub-group)
plt.figure(figsize=(15, 8))
cols = 3
rows = int(np.ceil(len(patches) / cols))

for i in range(len(patches)):

    plt.subplot(rows * 2, cols, i + 1 + (i // cols) * cols)
    plt.imshow(patches[i])
    plt.title(f"{patch_labels[i]}", fontsize=9)
    plt.axis("off")

    plt.subplot(rows * 2, cols, i + 1 + (i // cols + 1) * cols)
    for color, channel in zip(('red', 'green', 'blue'), (0, 1, 2)):
        hist, _ = np.histogram(patches[i][:, :, channel], bins=32, range=(0, 256))
        plt.plot(hist, color=color, alpha=0.7)
    plt.title(f"Hist P{i}", fontsize=8)
    plt.xlim([0, 32])

plt.tight_layout()
plt.suptitle("Golden Ratio Decomposition: Patches & Color Histograms", y=1.02, fontsize=15)
plt.show()


euclidean_dist_matrix = squareform(pdist(features_array, metric='euclidean'))


correlation_matrix = np.corrcoef(features_array)
correlation_dissimilarity = 1 - correlation_matrix

fig, axes = plt.subplots(1, 2, figsize=(15, 6))
labels = [f"P{i}" for i in range(len(patches))]


sns.heatmap(
    euclidean_dist_matrix,
    annot=True,
    fmt=".2f",
    cmap="YlOrRd",
    ax=axes[0],
    xticklabels=labels,
    yticklabels=labels
)
axes[0].set_title("Euclidean Distance Matrix (Golden Ratio Patches)")

sns.heatmap(
    correlation_dissimilarity,
    annot=True,
    fmt=".2f",
    cmap="Blues",
    ax=axes[1],
    xticklabels=labels,
    yticklabels=labels
)
axes[1].set_title("Correlation Dissimilarity Matrix (1 - Correlation)")

plt.tight_layout()
plt.show()
```
